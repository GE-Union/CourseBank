```python
# ---- Import af pakker ----
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
# ---- Definition af variable ----
l = 1 # m
g= 9.82 # m/(s^2)

# ---- startbetingelser ---- 
init = [np.pi/4,0] # første element er startbetingelsen for theta og andet element er startbetingelsen for omega
# ---- lister af punkterne ----
t_eval = np.linspace(0,20,1000)

# ---- definition af vores førsteordens differentialligninger
def f(t,y):
    theta,omega = y
    return [omega,-(3*g)/(2*l)*np.sin(theta)] #return the time derivatives [d(theta)/dt, d(omega)/dt]

# ---- definition af vores event ---- 
def maxima(t,y):
    theta,omega=y
    return omega #den tjekker for når omega = 0

maxima.direction = -1    # only maxima (omega: + -> -)
maxima.terminal = False

# ---- simulering ---- #
sol = solve_ivp(f,[t_eval[0],t_eval[-1]],init,t_eval=t_eval,events=[maxima])

# ---- display af løsningen ----
t = sol.t
x = sol.y[0]
plt.plot(t,x)
plt.xlabel('Time $t$ (s)')
plt.ylabel('theta (rad)')
plt.show()
print(sol.t_events[0][1],sol.t_events[0][2],sol.t_events[0][3]-sol.t_events[0][1])
```
