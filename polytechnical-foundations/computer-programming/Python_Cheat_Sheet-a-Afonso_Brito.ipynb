```python
#example of a class
class Dog:
    def __init__(self, name, age, race):
        self.name = name
        self.age = age
        self.race = race
    
    def make_n_years_older(self, num_years):
        self.age = self.age + num_years
    
    def get_description(self):
        description = f"{self.name} is a {self.age} years old {self.race}."
        return description
    
my_dog = Dog('Fiddo', 2, 'Border Collie')
print(my_dog.get_description()) #prints Fiddo is a 2 years old Border Collie.
my_dog.make_n_years_older(5)
print(my_dog.get_description()) #prints Fiddo is a 7 years old Border Collie.

#example of mutability
class Circle:
    def __init__(self, center, radius):
        self.center = center
        self.radius = radius

first_circle = Circle((0, 0), 5)
second_circle = first_circle
print(second_circle.radius) #prints 5

first_circle.radius = 10
print(second_circle.radius) #prints 10

#class naming

#CamelCase each word starts off with uppercase, should be used for ClassNames

#underscore_case each word is separated by '_', should be used for methods and attribute names

class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    def get_full_name(self, ):
        return self.first_name + " " + self.last_name
```

```python
#Inheritancen example

class NameTracker:

    def __init__(self):
        self.longest = ''
        self.shortest = ''
    
    def add(self, name):
        if len(name) > len(self.longest):
            self.longest = name
        if len(name) < len(self.shortest) or self.shortest=='':
            self.shortest = name
    
    def print(self):
        print('Names tracked so far')
        print('  Longest :', self.longest)
        print('  Shortest:', self.shortest)

class CarefulNameTracker(NameTracker):
    
    def add(self, name):
        if not (' ' in name or '-' in name):
            if len(name) > len(self.longest):
                self.longest = name
            if len(name) < len(self.shortest) or self.shortest=='':
                self.shortest = name

#CarefulNameTracker is a subclass (or child class) of NameTracker (superclass or parent class)

#the sub class could be rewritten as:

class CarefulNameTracker(NameTracker):
    def add(self, name):
        if not (' ' in name or '-' in name):
            super().add(name)        

#special methods

class Time:
    
    def __init__(self, hours, minutes):
        self.hours = hours
        self.minutes = minutes
    
    def increment_hours(self, ):
        self.hours += 1
        if self.hours == 24:
            self.hours = 0
    
    def increment_minutes(self, ):
        self.minutes += 1
        if self.minutes == 60:
            self.minutes = 0
            self.increment_hours()
    
    def __str__(self):
        return f"{self.hours:02d}:{self.minutes:02d}"
    
#operator overloading
class Time:
    
    def __init__(self, hours, minutes):
        self.hours = hours
        self.minutes = minutes
    
    def increment_hours(self, ):
        self.hours += 1
        if self.hours == 24:
            self.hours = 0
    
    def increment_minutes(self, ):
        self.minutes += 1
        if self.minutes == 60:
            self.minutes = 0
            self.increment_hours()
    
    def __str__(self):
        return f"{self.hours:02d}:{self.minutes:02d}"
    
    #let's us compare if the results are equal
    
    def __eq__(self, other):
        return (self.hours == other.hours) and (self.minutes == other.minutes)
    
    #let's us add the results correctly
    
    def __add__(self, other):
        minutes = self.minutes + other.minutes
        hours = self.hours + other.hours
        if minutes >= 60:
            minutes = minutes - 60
            hours = hours + 1
        if hours >= 24:
            hours = hours - 24
        return Time(hours, minutes)
    
```
