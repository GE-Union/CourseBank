### The binomial distribution 

The **Binomial distribution** models the number of successes in a fixed number of independent trials, each with the same probability of success.

Let:

- \( n \) be the number of independent trials  
- \( p \) be the probability of success in each trial  
- \( x \) be the number of observed successes  

The **probability mass function (PMF)** is given by:

$
f(x; n, p) = P(X = x) = \binom{n}{x} p^x (1 - p)^{n - x}
$

where the binomial coefficient is:

$
\binom{n}{x} = \frac{n!}{x!(n - x)!}
$

The probability using the pdf ("pmf") of the binomial: 
(pmf - pdf for discrete distributions)

### Hypergeometric distribution

The **Hypergeometric distribution** models the probability of drawing a certain number of successes when sampling **without replacement** from a finite population.

Let:

- $N$ = total population size  
- $a$ = number of successes in the population  
- $n$ = number of draws  
- $x$ = number of observed successes  

The **probability mass function (PMF)** is given by:

$
f(x; n, a, N) = P(X = x) = \frac{\binom{a}{x} \binom{N - a}{n - x}}{\binom{N}{n}}
$

where the binomial coefficient is defined as:

$
\binom{a}{x} = \frac{a!}{x!(a - x)!}
$

k - number of success, n - number of tries, p - percentage

The probability using the pdf ("pmf") of the hypergoemetrical distribution:

In the book we use `n` draws with `a` number of successes from a `N` size population. `(n,a,N)`. Whereas the corresponding Python (scipy.stats) arguments (parameters) would be `(N,n,M)` - so    `N` for number of draws, `n` number of successes from a `M` size population.

Additionally, `k` correponds to `x` from the book. - number of successes

```python
1 - stats.hypergeom.pmf(k=0, M=10, n=2, N=3)
```

Mean and variance of a hypergeometric distributed random variable:

$
\mu = n \cdot \frac{a}{N}
$

$
\sigma^2 = n \cdot \frac{a}{N} \cdot \frac{N - a}{N} \cdot \frac{N - n}{N - 1}
$

### t-distribution

Let:

- $\nu$ be the degrees of freedom  
- $\Gamma(\cdot)$ be the Gamma function  

**Probability Density Function (PDF):**

$
f_T(t) = \frac{\Gamma\left( \frac{\nu + 1}{2} \right)}{\sqrt{\nu \pi} \, \Gamma\left( \frac{\nu}{2} \right)} \left(1 + \frac{t^2}{\nu} \right)^{-\frac{\nu + 1}{2}}
$



**Relation to Normal and Chi-Square Distributions**

Let:

- $Z \sim N(0, 1)$  
- $Y \sim \chi^2(\nu)$  

Then:

$
X = \frac{Z}{\sqrt{Y/\nu}} \sim t(\nu)
$
**Sample-based t-Statistic**

For normal distributed random variables $X_1, \ldots, X_n$:

- $\overline{X}$ is the sample mean  
- $\mu$ is the population mean  
- $S$ is the sample standard deviation  
- $n$ is the sample size  

Then the t-statistic is:

$
T = \frac{\overline{X} - \mu}{S / \sqrt{n}} \sim t(n - 1)
$



**Mean and Variance of a t-Distributed Variable**

- Mean: $\mu = 0$ for $\nu > 1$  
- Variance: $\sigma^2 = \frac{\nu}{\nu - 2}$ for $\nu > 2$

### Chi-Square ($\chi^2$) Distribution

Let:

- $\nu$ be the degrees of freedom  
- $\Gamma(\cdot)$ be the Gamma function  

**Probability Density Function (PDF):**

$
f(x) = \frac{1}{2^{\nu/2} \, \Gamma\left( \frac{\nu}{2} \right)} x^{\frac{\nu}{2} - 1} e^{-x/2}, \quad x \geq 0
$


**Relation to Sample Variance**

Given a sample of size $n$ from normally distributed random variables $X_i$ with variance $\sigma^2$, the sample variance $S^2$ can be used to form a $\chi^2$-distributed variable:

$
\chi^2 = \frac{(n - 1) S^2}{\sigma^2}
$

with degrees of freedom $\nu = n - 1$.



**Mean and Variance**

  $

- **Variance:**  
  $

  $

### Q-Q plot

The simplest observations in a Q-Q plot before conducting a hypothesis test using a small dataset are:

1. **Linearity of Points**: Check if the points fall approximately along a straight line. This indicates that the data is consistent with the theoretical distribution (e.g., normal).

2. **Systematic Deviations**: Look for any systematic patterns, such as curves or S-shapes, which suggest the presence of skewness in the data.

3. **Ends of the Plot**: Observe if the points at the ends deviate significantly from the line, indicating heavy tails or kurtosis (extreme values more frequent than expected).

4. **Outliers**: Identify any points that fall far away from the reference line, which could be potential outliers in the dataset.

These basic observations help determine if the data distribution is approximately normal or has deviations that should be considered before performing hypothesis testing.

### One-way ANOVA

#### *Block* Analogy Explanation:

Think of ANOVA like applying **treatment** fertilizers in into different "**blocks**" of fields.<br>
We want to know **if the fertilizer or treatment significantly impacts crop yield from the blocks**<br>
Variation WITHIN each fertilizer or *treatment* groups vs. variation BETWEEN fertilizer or *treatment* groups

#### Null Hypothesis ($H_0$):
  $
    \mu_1 = \mu_2 = \mu_3
    $
- The mean corn yields are equal across all three fertilizer types.
- There is no significant difference in crop yield between the Organic, Chemical, and Balanced fertilizers.

#### Alternative Hypothesis ($H_1$):
$
  \text{At least one } \mu \text{ differs}
  $
- At least one fertilizer type has a significantly different mean yield.
- Not all fertilizer group means are the same.

#### Symbolic Representation:
$  H_0: \mu_\text{Organic} = \mu_\text{Chemical} = \mu_\text{Balanced}$ <br>
$H_1: \text{At least one } \mu \text{ is different from the others.}$

#### Decision Rule:
- Reject $H_0$ if $p < 0.05$: Fertilizer types significantly differ in mean yield.
- Fail to reject $H_0$ if $p \geq 0.05$: No significant evidence of differences in mean yield.

| Source of variation | Degrees of freedom     | Sums of squares        | Mean sum of squares                                | Test-statistic \( F \)                    | *p*-value                  |
|---------------------|------------------------|-------------------------|----------------------------------------------------|--------------------------------------------|----------------------------|
| Treatment           | $$k - 1$$              | $$SS(\text{Tr})$$       | $$MS(\text{Tr}) = \frac{SS(\text{Tr})}{k - 1}$$    | $$F_{\text{obs}} = \frac{MS(\text{Tr})}{MSE}$$ | $$P(F > F_{\text{obs}})$$ |
| Residual            | $$n - k$$              | $$SSE$$                 | $$MSE = \frac{SSE}{n - k}$$                         |                                            |                            |
| Total               | $$n - 1$$              | $$SST$$                 |                                                    |                                            |                            |


```python
f_statistic, p_value = stats.f_oneway(first_treatment, second_treatment, third_treatment)

print("ANOVA Results:")
print(f"F-statistic: {f_statistic}")
print(f"P-value: {p_value}")

# Interpretation
alpha = 0.05
if p_value < alpha:
    print("Fertilizer types as treatements significantly differ the crop yield. So, at least one mean is differnt from others")
else:
    print("No significant difference between fertilizer types")
```

computing SSE (sum of squared errors) and MSE

$

```python
SSTr = SST - SSE
MSTr = SSTr / (k-1)
```

#### Model control
The assumptions for one-way ANOVA model are:
1. The data comes from a normal distribution in each group
2. The variances from each group are the same

The assumptions can be controlled by:
1. The normality within groups assumption: 
   1.  q-q plot within each group but exercise data as low as four observations are not insightful.
   2.  All the residuals for all groups are pooled in a QQ plot, that is, they should all follow the same zero-mean normal distribution.
2.  The homogeneous variances assumption can be controlled by simply looking at the distributions within each sample, most conveniently for this purpose by the group-wise box plot.
